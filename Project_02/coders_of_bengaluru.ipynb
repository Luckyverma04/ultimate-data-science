```python
with open("initialdata.txt", encoding='utf-8') as f:
    data = f.read()
```

```python
chunks = data.split("\n\n") 
chunks = [c for c in chunks if len(c)>3]
```

```python

def parse_chunk(chunk):
    try:
        chunk = chunk.strip()
        sep_chunk = chunk.split('\n')

        username = sep_chunk[0]

        no_of_post = int(
            sep_chunk[2].split("post")[0].replace(",", "")
        )

        no_of_followers = float(
            sep_chunk[3].split("follower")[0]
            .replace(",", "").replace("K", "").replace("k", "").replace("M", "")
        )

        if "K" in sep_chunk[3] or "k" in sep_chunk[3]:
            no_of_followers = int(no_of_followers * 1000)

        elif "M" in sep_chunk[3]:
            no_of_followers = int(no_of_followers * 1000000)
        else:
            no_of_followers = int(no_of_followers)

        
        no_of_following = float(
            sep_chunk[4].split("following")[0]
            .replace(",", "").replace("K", "").replace("k", "").replace("M", "")
        )

        if "K" in sep_chunk[4] or "k" in sep_chunk[4]:
            
            no_of_following = int(no_of_following  * 1000)

        elif "M" in sep_chunk[4]:
            no_of_following  = int(no_of_following  * 1000000)
        else:
            no_of_following  = int(no_of_following )


            
        name = sep_chunk[1]

        if len(sep_chunk) > 5:
            type_of_pages = sep_chunk[5]
            bio = "\n".join(sep_chunk[6:])
        else:
            type_of_pages = "unknown"
            bio = ""

        return {
            "username": username,
            "no_of_post": no_of_post,
            "no_of_followers": no_of_followers,
            "no_of_following": no_of_following,
            "name": name,
            "type_of_pages": type_of_pages,
            "bio": bio
        }

    except Exception as e:
        print(e, chunk)
        return None


all_chunks = []

for chunk in chunks:
    parsed_chunk = parse_chunk(chunk)

    if parsed_chunk is not None:
        all_chunks.append(parsed_chunk)
```

```python
all_chunks = []
for chunk in chunks:
    parsed_chunk = parse_chunk(chunk)
    all_chunks.append(parsed_chunk)
```

```python
import json
s=json.dumps(all_chunks,indent = 4)
with open("final.json","w")as f:
    f.write(s)
```

```python
# count of creators types:
count_category = {}
for chunk in all_chunks:
    category = chunk['type_of_pages']
    count_category[category] =  count_category.get(category, 0) + 1
print(count_category)
print("Total categories:", len(count_category))
```
