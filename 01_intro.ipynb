```python
import json
```

```python
#this function will load the data:
def load_data(filename):
    with open(filename,"r")as f:
        data = json.load(f)
    return data 
    
```

```python
data = load_data("data.json")
```

```python
#write a function to display users and their connection
def display_user(data):
    print("user and their connections\n")
    for user in data['users']:
        print(f"ID:{user['id']}.{user['name']} is friend with:{user['friends']} and liked pages are {user['liked_pages']}")
    print("\npages information")
    for page in data['pages']:
        print(f"{page['id']}: {page['name']}")
display_user(data)       
```
