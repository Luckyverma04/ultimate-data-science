```python
import pandas as pd

# Sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Math': [85, 78, 92],
    'Science': [90, 82, 89],
    'English': [88, 85, 94]
}

df = pd.DataFrame(data)

# Display the DataFrame
print(df)
```

```python
df2 = df.melt(id_vars=["Name"], value_vars=["Math", "Science", "English"], var_name="Subject", value_name="Score")
df2
```

```python
df2 = print(df2.columns.tolist())
```

```python
# Step 1: Create df2 again
df2 = df.copy()

# Step 2: Check columns
print(df2.columns.tolist())

# Step 3: Pivot
df2 = df2.pivot(
    index="Name",
    columns="Subject",
    values="Score"
)

# Step 4: Display
df2
```
